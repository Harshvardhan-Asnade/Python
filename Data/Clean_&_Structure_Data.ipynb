```python
import json
```

```python
def load_data(filename): 
    with open(filename,'r') as f: 
        data=json.load(f)

    return data
```

```python
data =load_data("Example_store_data ")
print(data)
```

```python
word_to_num = {
    "one": 1,
    "two": 2,
    "three": 3,
    "four": 4,
    "five": 5
}
```

```python

def clean_rating(value):
    if value is None:
        return None
    value = str(value).lower()
    if value in word_to_num:
        return word_to_num[value]
    return float(value)

for row in data:
    row["rating"] = clean_rating(row.get("rating"))
```

```python
ages = []
for row in data:
    if "age" in row:
        ages.append(int(row["age"]))

avg_age = sum(ages) / len(ages)

for row in data:
    if "age" not in row:
        row["age"] = avg_age
    else:
        row["age"] = int(row["age"])
```

```python
unique_data = []
seen = set()

for row in data:
    identifier = (row["name"], row["age"])
    if identifier not in seen:
        seen.add(identifier)
        unique_data.append(row)

data = unique_data
```

```python

total = 0
count = 0

for row in data:
    total += row["rating"]
    count += 1

avg_rating = total / count

poor_count = 0
for row in data:
    if row["rating"] < 3:
        poor_count += 1

poor_percentage = (poor_count / len(data)) * 100
```

```python
for row in data:
    if row["rating"] >= 4:
        row["recommendation"] = "Same brand"
    else:
        row["recommendation"] = "Different brand"
```

```python
print("Cleaned Data:\n")
for row in data:
    print(row)

print("\nAverage Rating:", avg_rating)
print("Poor Rating %:", poor_percentage)
```
